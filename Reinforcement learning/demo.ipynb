```python
import numpy as np

# Environment: Room with dust patches
def simulate_vacuum(p_forward, r_angle, steps=300):
    dust_collected = 0
    position = np.array([0, 0])  # start at origin
    direction = 0  # facing right (0 degrees)

    for _ in range(steps):
        if np.random.rand() < p_forward:
            # move forward
            position.__add__(np.array([np.cos(np.deg2rad(direction)), np.sin(np.deg2rad(direction))]))
            # position += np.array([np.cos(np.deg2rad(direction)), np.sin(np.deg2rad(direction))])
        else:
            # rotate randomly
            direction += np.random.uniform(-r_angle, r_angle)

        # reward: dust collected (random patches)
        if np.random.rand() < 0.05:  # 5% chance of dust at new position
            dust_collected += 1

    return dust_collected

# Brute force policy search
best_reward = -1
best_params = None

for p in [0.3, 0.5, 0.7]:
    for r in [30, 60, 90]:
        reward = simulate_vacuum(p, r)
        print(f"Policy (p={p}, r={r}) -> Dust collected: {reward}")
        if reward > best_reward:
            best_reward = reward
            best_params = (p, r)

print("Best policy:", best_params, "with reward:", best_reward)
```
