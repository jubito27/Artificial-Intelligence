Xavior Glorat - Specificially used when activation function is tanh

```python
from sklearn.datasets import make_moons


x , y = make_moons(100 , noise=0.3 , random_state=1)
```

```python
import tensorflow
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
model1 = Sequential()

model1.add(Dense(128 , activation='tanh' , kernel_initializer='glorot_uniform' )) #By default is also glorot Unifom
model1.add(Dense(64 , activation='tanh' , kernel_initializer='glorot_uniform' ))
model1.add(Dense(32 , activation='tanh' , kernel_initializer='glorot_uniform' ))
model1.add(Dense(16 , activation='tanh' , kernel_initializer='glorot_uniform' ))
model1.add(Dense(8 , activation='tanh' , kernel_initializer='glorot_uniform' ))
model1.add(Dense(4 , activation='tanh' , kernel_initializer='glorot_uniform' ))
model1.add(Dense(1 , activation='sigmoid'  ))

model1.compile(loss='binary_crossentropy' , optimizer='Adam' , metrics=['accuracy'])
model1.fit(x , y , epochs=100 , validation_split=0.2)
```

```python
from mlxtend.plotting import plot_decision_regions
import matplotlib.pyplot as plt
plot_decision_regions(x , y.astype('int') , clf=model1 , legend=2)
plt.xlim(-2 , 3)
plt.ylim(-1.5 , 2)
plt.show()
```
