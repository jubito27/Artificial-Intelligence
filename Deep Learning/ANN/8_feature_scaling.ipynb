```python
import pandas as pd
df = pd.read_csv(r"C:\Users\Abhishek sharma\Artificial Intelligence\Datasets\Social_Network_Ads.csv")
df.head()
```

```python
df = df.iloc[:,2:]
df.head()
```

```python
x = df.iloc[:,:-1]
x.head()
```

```python
y = df.iloc[:,-1]
y.head()
```

```python
import sklearn
```

```python
import sklearn.model_selection


x_train , x_test , y_train , y_test = sklearn.model_selection.train_test_split(x , y , test_size=0.3 , random_state=0)
```

```python
import sklearn.preprocessing

scale = sklearn.preprocessing.StandardScaler()
```

```python
x_train_scaled = scale.fit_transform(x_train)
```

```python
import numpy


std = numpy.round(x_train_scaled.std())
mean = numpy.round(x_train_scaled.mean())
print("standard Deviation : " , std)
print("Mean : " , mean)
```

```python
x_test_scaled = scale.transform(x_test)
```

```python
std = numpy.round(x_test_scaled.std())
mean = numpy.round(x_test_scaled.mean())
print("standard Deviation : " , std)
print("Mean : " , mean)
```

```python
numpy.mean(x_test_scaled)
```

```python
import tensorflow
```

```python
model = tensorflow.keras.models.Sequential()
```

```python
import keras


model.add(keras.layers.Dense(16 , activation='relu'))
model.add(keras.layers.Dense(1 , activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy' , optimizer='adam' , metrics=['accuracy'])
```

```python
history = model.fit(x_train_scaled, y_train , validation_data=[x_test_scaled , y_test] , epochs=100)
```
