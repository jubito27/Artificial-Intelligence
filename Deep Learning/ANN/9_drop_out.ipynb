```python
import pandas as pd
df = pd.read_csv(r"C:\Users\Abhishek sharma\Artificial Intelligence\Datasets\Social_Network_Ads.csv")
df.head()
```

```python
x = df.drop(columns=['Gender' , 'Purchased'])
y = df.iloc[:,-1]
```

```python
import sklearn
import sklearn.model_selection


x_train , x_test , y_train , y_test = sklearn.model_selection.train_test_split(x , y , test_size=0.2 , random_state=0)
```

```python
x_train.shape , x_test.shape
```

```python
import sklearn.preprocessing


scaler = sklearn.preprocessing.StandardScaler()
```

```python
import numpy


x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
numpy.round(x_test_scaled.mean())
```

```python
import keras


model = keras.models.Sequential()
```

```python
model.add(keras.layers.Dense(32 , activation='relu'))
model.add(keras.layers.Dropout(0.5))
model.add(keras.layers.Dense(16 , activation='relu'))
model.add(keras.layers.Dropout(0.5))
model.add(keras.layers.Dense(8 , activation='relu'))
model.add(keras.layers.Dropout(0.5))
model.add(keras.layers.Dense(1 , activation='sigmoid'))
```

```python
model.compile(loss='binary_crossentropy' , optimizer='adam' , metrics=['accuracy'])
```

```python
model.fit(x_train_scaled , y_train , validation_data=[x_test_scaled , y_test] , epochs=100 , verbose=1)
```
