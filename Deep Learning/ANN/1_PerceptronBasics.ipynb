```python
import numpy as np
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt
```

```python
cgpa = [8.14 , 6.17 , 8.24 , 6.33 , 7.55]
resume = [6.52 , 5.28 , 8.24 , 7.27 , 7.30]
placed = [1 , 0 , 1 ,1 ,1]

data = {
    "cgpa" : cgpa,
    "resume_score" : resume,
    "placement" : placed
}

df = pd.DataFrame(data)

np.array(df.iloc[0:5 , 2:3]).transpose()
```

```python
sb.scatterplot(np.array(df.iloc[0:5 , 0:1]).transpose() , np.array(df.iloc[0:5 , 1:2]).transpose() , hue=np.array(df.iloc[0:5 , 2:3]).transpose())
#plt.scatter(df['cgpa'] ,df['resume_score']  ,df['placement'])
```

```python
x = df.iloc[: , 0:2]
y = df.iloc[: , -1]
```

```python
from sklearn.linear_model import Perceptron
```

```python
percep = Perceptron()
```

```python
percep.fit(x ,y)
```

```python
percep.coef_
```

```python
percep.intercept_
```

```python
from mlxtend.plotting import plot_decision_regions
```

```python
plot_decision_regions(x.values , y.values , clf=percep , legend=2) 
```
