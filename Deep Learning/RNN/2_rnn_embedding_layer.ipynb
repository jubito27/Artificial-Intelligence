Embedding layer is used to converting textual data into a Dense Matrix with its semantic meaning...

```python
import keras
from keras.datasets import imdb
from keras.utils import pad_sequences
from keras import Sequential
from keras.layers import SimpleRNN , Dense , Embedding , Flatten , Embedding
```

```python
(x_train , y_train) ,( x_test , y_test) = imdb.load_data()
```

```python
x_train = pad_sequences(x_train , padding='post' , maxlen=50)
x_test = pad_sequences(x_test , padding='post' , maxlen=50)
```

```python
x_train.shape
```

```python
model = Sequential()

model.add(Embedding(10000 , 2)) # Embedding layer is added at first with no of inputs and suze of vector to represent the data
model.add(SimpleRNN(32 , return_sequences=False ))
model.add(Dense(1 , activation='sigmoid'))
model.summary()
```

```python
model.compile(loss='binary_crossentropy' , optimizer='adam' , metrics = ['accuracy'])
history = model.fit(x_train , y_train , epochs=5 , validation_data=(x_test , y_test))
```
