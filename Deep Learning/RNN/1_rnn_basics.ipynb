RNN - Recurrent Neural Network... Used to work on text data ... this can easily understand the patterns of textual data without loosing its meaning

```python
from keras.datasets import imdb
from keras.utils import pad_sequences
from keras import Sequential
from keras.layers import SimpleRNN , Dense , Embedding , Flatten
```

```python
(x_train , y_train) ,( x_test , y_test) = imdb.load_data()
```

```python
x_train = pad_sequences(x_train , padding='post' , maxlen=100)
x_test = pad_sequences(x_test , padding='post' , maxlen=100)
```

```python
model = Sequential()

model.add(SimpleRNN(32 , input_shape =(100 ,1) , return_sequences=False))
model.add(Dense(1 , activation='sigmoid'))
model.summary()
```

```python
model.compile(loss='binary_crossentropy' , optimizer='adam' , metrics=['accuracy'])
model.fit(x_train , y_train , epochs=5 , validation_data=(x_test , y_test))
```
