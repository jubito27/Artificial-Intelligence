```python
import keras
from keras import datasets
```

```python
(x_train, y_train), (x_test, y_test) = datasets.mnist.load_data()
```

```python
from keras import Sequential
from keras.layers import Dense , Flatten , Conv2D , AvgPool2D
```

LeNet Architecture - First CNN architecture

```python
model = Sequential()

model.add(Conv2D(6 , kernel_size=(5,5) , padding='valid' , activation='tanh' , input_shape=(32,32,1)))
model.add(AvgPool2D(pool_size=(2,2) , strides=2 , padding='valid'))

model.add(Conv2D(16 , kernel_size=(5,5) , padding='valid' , activation='tanh'))
model.add(AvgPool2D(pool_size=(2,2) , strides=2 , padding='valid'))

model.add(Flatten())

model.add(Dense(120 , activation='tanh'))
model.add(Dense(84 , activation='tanh'))
model.add(Dense(10 , activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(optimizer='Adam' , loss='binarysparse_categorical_crossentropy' , metrics=['accuracy'])
model.fit(x_train , y_train , epochs=100 , validation_data=(x_test , y_test))
```
