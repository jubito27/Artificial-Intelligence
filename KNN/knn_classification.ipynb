```python
import numpy as np
from collections import Counter
class MyKNNClass:

    def __init__(self , k):
        self.k = k
        self.x = None
        self.y = None

    def fit(self, x , y):
        self.x = x
        self.y = y

    def distances(self ,a , b):
        return np.linalg.norm(a-b)

    def majority(self , neibours):
        votes = []
        for i in neibours:
            votes.append(self.y[i[0]])
        votes = Counter(votes)
        return votes.most_common()[0][0]

    def predict(self , x):
        y_pred = []
        for i in x:
            distance = []
            for j in self.x:
                distance.append(self.distances(i , j))
            sortedDistance = sorted(list(enumerate(distance)) , key=lambda x:x[1])[0:self.k]
        
            label = self.majority(sortedDistance)
            y_pred.append(label)
        return np.array(y_pred)

                
```

```python
my = MyKNNClass(k=3)
my.fit(x_train , y_train)
sd = my.predict(x_test)
```

```python
accuracy_score(y_test , sd)
```
