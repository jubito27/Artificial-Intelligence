```python
class MyLinearregression:
    def __init__(self):
        self.m = []
        self.b = None
    
    def fit(self , x_train , y_train):
        
        x = np.array(x_train)
        y = np.array(y_train)
        num = 0
        den = 0
        shape = x_train.shape
        for i in range(shape[1]):
            for j in range(shape[0]):
                if len(x) == len(y):
                    num += (x[j]-x.mean())*(y[j]-y.mean())
                    den += (x[j]-x.mean())*(x[j]-x.mean())
                else:
                    print(f"Give both x_train that is {len(x)} and y_train that is {len(y)} of equal size")
            div = num/den
            m_list = self.m
            m_list.append(div)
        self.m = np.array(self.m)


        self.b = y.mean() - (self.m.mean()*x.mean())


    def predict(self, x_test):
            x = np.array(x_test)
            y_pred = []
            for i in range(x.shape[0]):
                pred = np.dot(self.m, x[i]) + self.b
                y_pred.append(pred)
            return np.array(y_pred)
```

```python
start_time = time.perf_counter_ns()
lr = MyLinearregression()
lr.fit(x_train  , y_train)
y_predict = lr.predict(x_test=x_test)
final_time = time.perf_counter_ns() - start_time


```
