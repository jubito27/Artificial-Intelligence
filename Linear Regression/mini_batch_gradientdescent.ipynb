```python
from sklearn.datasets import load_diabetes

x , y = load_diabetes(return_X_y=True)
from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x , y, test_size=0.2 , random_state=42)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train , y_train)

y_pred = model.predict(x_test)

from sklearn.metrics import r2_score

r2_score(y_test , y_pred)
```

```python
model.coef_ , model.intercept_
```

```python
import numpy as np
import random as rd
class MBGDRegressor:
        def __init__(self ,batch_size , epochs = 10 , lr= 0.1 ):
            self.coef_ = None
            self.intercept_ = None
            self.epochs = epochs
            self.lr = lr
            self.batch_size = batch_size

        def fit(self , x ,y  ,  verbose = False):
            if self.batch_size > len(x):
                 print("Batch size cant be greater than " , len(x))
            batches = len(x)//self.batch_size
          
            if self.epochs <= x.shape[0]:
                self.intercept_ = 0
                self.coef_ = np.ones(x.shape[1])
                for i in range(self.epochs):
                    for j in range(x.shape[0]//batches):
                        if verbose:
                            print(f"Coefficients at {i} loop: {self.coef_}")
                            print(f"Intercept at {i} loop: {self.intercept_}")
                            print("-"*50)
                        idx = rd.sample(range(x.shape[0]) , x.shape[1])
                        y_cap = np.dot(x[idx] , self.coef_) + self.intercept_
                        i_slope = -2 * np.sum(y[idx] - y_cap)
                        c_slope = -2 * np.dot((y[idx]  - y_cap), x[idx])

                        self.intercept_ = self.intercept_ - (self.lr * i_slope)
                        self.coef_ = self.coef_ - (self.lr * c_slope)
                print("Model trained successfully")
            else:
                 print("Epochs size cant be greater than " , x.shape[0])

        def predict(self , x):
             y_pred= np.dot( x ,self.coef_) + self.intercept_
             return y_pred   
```

```python
my = MBGDRegressor(batch_size=20 , epochs=50)
my.fit(x_train , y_train )
```

```python
my.coef_ , my.intercept_ 
```

```python
my_pred = my.predict(x_test)
r2_score(y_test , my_pred)
```
