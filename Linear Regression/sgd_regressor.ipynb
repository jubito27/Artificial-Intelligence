```python
from sklearn.datasets import load_diabetes

x , y = load_diabetes(return_X_y=True)
from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x , y, test_size=0.2 , random_state=42)
```

```python
from sklearn.linear_model import SGDRegressor

sgd = SGDRegressor()

sgd.fit(x_train , y_train)
```

```python
y_pred = sgd.predict(x_test)

from sklearn.metrics import r2_score
r2_score(y_test , y_pred)
```

```python
sgd.intercept_ , sgd.coef_
```

```python
import numpy as np
class MySGDRegressor:
        def __init__(self , epochs = 10 , lr= 0.1):
            self.coef_ = None
            self.intercept_ = None
            self.epochs = epochs
            self.lr = lr

        def fit(self , x ,y , verbose = False)  :
            if self.epochs <= x.shape[0]:
              self.intercept_ = 0
              self.coef_ = np.ones(x.shape[1])
              for i in range(self.epochs):

                    if verbose:
                        print(f"Coefficients at {i} loop: {self.coef_}")
                        print(f"Intercept at {i} loop: {self.intercept_}")
                        print("-"*50)

                    y_cap = np.dot(x[i] , self.coef_) + self.intercept_
                    i_slope = -2 * (y[i] - y_cap)
                    c_slope = -2 * np.dot((y[i]  - y_cap), x[i])

                    self.intercept_ = self.intercept_ - (self.lr * i_slope)
                    self.coef_ = self.coef_ - (self.lr * c_slope)
            else:
                 print("Epochs size cant be greater than " , x.shape[0])

        def predict(self , x):
             y_pred= np.dot( x ,self.coef_) + self.intercept_
             return y_pred   
```

```python
my = MySGDRegressor(epochs=353 , lr = 0.08)
my.fit(x_train, y_train , verbose=True)
```

```python
import numpy as np
class SGDRegressor:
        def __init__(self , epochs = 10 , lr= 0.1):
            self.coef_ = None
            self.intercept_ = None
            self.epochs = epochs
            self.lr = lr

        def fit(self , x ,y , verbose = False)  :
            if self.epochs <= x.shape[0]:
                self.intercept_ = 0
                self.coef_ = np.ones(x.shape[1])
                for i in range(self.epochs):
                    for j in range(x.shape[0]):
                        if verbose:
                            print(f"Coefficients at {i} loop: {self.coef_}")
                            print(f"Intercept at {i} loop: {self.intercept_}")
                            print("-"*50)
                        idx = np.random.randint(0 , x.shape[0])
                        y_cap = np.dot(x[idx] , self.coef_) + self.intercept_
                        i_slope = -2 * (y[idx] - y_cap)
                        c_slope = -2 * np.dot((y[idx]  - y_cap), x[idx])

                        self.intercept_ = self.intercept_ - (self.lr * i_slope)
                        self.coef_ = self.coef_ - (self.lr * c_slope)
                print("Model trained successfully")
            else:
                 print("Epochs size cant be greater than " , x.shape[0])

        def predict(self , x):
             y_pred= np.dot( x ,self.coef_) + self.intercept_
             return y_pred   
```

```python
mymodel = SGDRegressor(epochs=80 , lr=0.05)
mymodel.fit(x_train , y_train)
```

```python
mymodel_pred = mymodel.predict(x_test)
```
