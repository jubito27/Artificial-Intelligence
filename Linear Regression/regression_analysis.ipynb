```python
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = 10 * np.random.rand(50)
y= 2 * x-5 + np.random.rand(50)* 4
x = x.reshape(-1 , 1)

model = LinearRegression()
model.fit(x , y)

y_pred = model.predict(x)


plt.scatter(x , y , label="Data Points")
plt.xlabel("x")
plt.ylabel("y")
plt.title("SOmething Useless")


x_line = np.linspace(0 , 10 , 100)
y_actual = 2 * x_line - 5
plt.plot(x_line , y_actual , 'r' , label = "Population line (m=2 , y=-5)")

y_estimateed = model.coef_[0] * x_line + model.intercept_
plt.plot(x_line , y_estimateed , "g" , label = f" Estimated line (m={model.coef_[0]:.2f} , (b={model.intercept_:.2f}))")

plt.legend()
plt.show()
```
