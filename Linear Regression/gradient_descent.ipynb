```python
import numpy as np
class GradientRegressor:

    def __init__(self):
        self.m = 1
        self.b = 0
        self.lr = 0.1
        self.epochs = 100

    def fit(self ,x , y , verbose=False):
        m = self.m
        b= self.b
        lr_ =self.lr
        if len(x) == len(y):
            for i in range(self.epochs):
                if verbose:
                    print(f" b at {i} iteration : {b}")
                    print(f" m at {i} iteration : {m}")
                    print("-"*20)
                intercept_slope =  -2 * np.sum(y- m*x.ravel() - b)
                coef_slope = -(2)* np.sum(x.ravel()* (y - m*x.ravel() - b))
                b = b - lr_ * intercept_slope
                m = m - lr_ * coef_slope
            self.m = m
            self.b = b
            plt.scatter(x , y)
            plt.plot(x , self.predict(x) , color='red')
            plt.show()
            print("Training Completed Successfully")
        else:
            print("Length of x should be equal to length of y")

    def predict(self , x):
        y_pred = (self.m * x.ravel())  + self.b 
        return y_pred
```

```python
model = GradientRegressor()
model.fit(x , y , verbose=True)
```
