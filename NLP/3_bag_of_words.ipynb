```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
```

```python
data = {
    "text" : [
        "Naruto better onepiece",
        "onepiece better Bleach",
        "Bleach better AOT",
        "AOT better Naruto",
        "Blueloch better demonslayer",
        "demonslayer bluelock rezero demonslayer"
    ],
    "label" : [1 , 0 , 1, 1, 0 , 0]
}

df = pd.DataFrame(data)
df
```

```python
bow = CountVectorizer()
text_num = bow.fit_transform(df['text'])
```

```python
text_num.toarray()
```

```python
bow.vocabulary_
```

```python
bow.transform(["Naruto is a best anime of all time than one piece bleach AOT and other animes like Demonslayer bluelock"]).toarray()
```

```python
import numpy as np


def transform(text):
    voc = bow.vocabulary_
    ar = np.array([])
    for i in range(len(voc)):
        z = voc.values() == i
        if z in text:
            ar.insert(1)
        else:
            ar.insert(0)
    return ar

transform("Naruto is a best anime of all time than one piece bleach AOT and other animes like Demonslayer bluelock")
```
