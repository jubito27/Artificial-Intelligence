```python
from nltk.corpus import opinion_lexicon
import nltk

nltk.download('opinion_lexicon')
nltk.download('punkt')

```

```python
pos_words = set(opinion_lexicon.positive())
neg_words = set(opinion_lexicon.negative())
```

```python

def sentiment_analysis(text):
    words = text.lower().split()
    pos = 0
    neg = 0

    for word in words:

        if word in pos_words:
            pos += 1

        if word in neg_words:
            neg += 1

    print(f"positive words  : {pos} , Negative words : {neg}" )

    if pos > neg:
        return "Positive"

    elif neg > pos:
        return "Negative"

    else:
        return "Neutral"
```

```python
text = "The movie was great in many ways, but it also had some really bad moments that were hard to ignore. The performances by the lead actors were excellent and their chemistry made several scenes feel genuinely happy and heartfelt. I absolutely love how the director handled the emotional parts, and the soundtrack was awesome, adding a good amount of depth to key moments. However, the pacing of the story was poor in the middle, and a few subplots felt like the worst kind of filler that added nothing but confusion. By the end, some characters’ decisions made me a bit sad, and the unresolved issues created a lingering sense of problem with the overall narrative, so while parts of it were ok, it left me feeling both impressed and slightly disappointed."
sentiment_analysis(text)
```
