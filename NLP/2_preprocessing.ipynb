<span style="font-size:32px">Punctuations</span>

```python
import string
def remove_punctuations(text):
    if not isinstance(text, str):
        return text
    return text.translate(str.maketrans('' , '' , string.punctuation))
```

<span style="font-size:32px">Chat Word Conversion</span>

```python
# safely read the slang file (don't reuse a possibly-closed f)
with open('slang.txt', 'r', encoding='cp1252', errors='ignore') as fh:
    slang_text = fh.read()
```

```python
# convert the loaded slang_text (string) into a dict so chat_conversion can index it
slang_map = {}
for line in slang_text.splitlines():
    line = line.strip()
    if not line or '=' not in line:
        continue
    k, v = line.split('=', 1)
    slang_map[k.strip().upper()] = v.strip()

# overwrite slang_text (the name used inside chat_conversion) with the dict
slang_text = slang_map

# now call the function
```

```python
def chat_conversion(text):
    new_text = []
    for w in text.split():
        if w.upper() in slang_text:
            new_text.append(slang_text[w.upper()])
        else:
            new_text.append(w)
    return " ".join(new_text)
```

```python
chat_conversion("LMAO whhat a dream but idk as iykyk")
```

<span style="font-size:32px">Stop Words</span>

```python
def stop_words(text):
    from nltk.corpus import stopwords
    if not isinstance(text , str):
        return text
    new_text = []
    stopword = stopwords.words('english')
    for word in text.split():
        if word in stopword:
            new_text.append('')
        else:
            new_text.append(word)
    x = new_text[:]
    new_text.clear()
    return ' '.join(x)

stop_words("I are the words best ninja in a naurto world of anime")
```

```python
df['overview'].apply(stop_words)
```

<span style="font-size:32px">Removing Emojis</span>

```python
import re

def remove_emoji(text):
    if not isinstance(text, str):
        return text

    emoji_pattern = re.compile(
        "["

        # emoticons
        "\U0001F600-\U0001F64F"

        # symbols & pictographs
        "\U0001F300-\U0001F5FF"

        # transport & map symbols
        "\U0001F680-\U0001F6FF"

        # flags (iOS)
        "\U0001F1E0-\U0001F1FF"

        # dingbats
        "\U00002700-\U000027BF"

        # miscellaneous symbols
        "\U00002600-\U000026FF"

        # enclosed characters / other pictographs
        "\U000024C2-\U0001F251"

        # supplemental symbols & pictographs (covers 🥰, 🥱, 🤧, 🤦 etc.)
        "\U0001F900-\U0001F9FF"

        # Symbols & Pictographs Extended-A (newer emojis)
        "\U0001FA70-\U0001FAFF"

        # some other supplemental ranges (optional)
        "\U0001F700-\U0001F77F"

        "]+",
        flags=re.UNICODE
    )

    # remove matched emoji characters
    text = emoji_pattern.sub("", text)

    # also remove variation selectors and ZWJ (these are used in emoji sequences)
    text = re.sub(r"[\uFE0F\u200D]", "", text)

    return text

remove_emoji("he honey 💕😊🥰🥰🥰🤧😞😑🟣😖😤😭😘🥱😘😭😤😖🤧🥰🎙️😒😂🤣❤️👌😁👍🙌🤦‍♀️")
```

<span style="font-size:32px">Word Tokenisation</span>

```python
def word_tokenize(text):
    text = '' if text is None else str(text)
    try:
        import spacy
        try:
            nlp = spacy.load('en_core_web_sm')
        except OSError:
            # model not found -> try to download it (requires internet)
            from spacy.cli import download
            download('en_core_web_sm')
            nlp = spacy.load('en_core_web_sm')

        doc = nlp(text)
        # return a list of token strings for simplicity
        return [token.text for token in doc]

    except Exception:
        # fallback to NLTK tokenizer if spaCy isn't available or download fails
        import nltk
        nltk.download('punkt', quiet=True)
        return nltk.word_tokenize(text)

# example
word_tokenize("I am ph.d in A.I")
```

<span style="font-size:32px">Sentence TOkenization</span>

```python
def sentence_tokenize(text):
    """Return list of sentences for `text` using spaCy when available, otherwise fall back to NLTK.

    - Tries to load `en_core_web_sm` and download it if missing.
    - Adds spaCy's `sentencizer` if not present (lightweight sentence boundary detector).
    - Falls back to `nltk.sent_tokenize` (ensures `punkt` is downloaded) if spaCy isn't available.
    """
    text = '' if text is None else str(text)
    try:
        import spacy
        try:
            nlp = spacy.load('en_core_web_sm')
        except OSError:
            # model not found -> try to download it (requires internet)
            from spacy.cli import download
            download('en_core_web_sm')
            nlp = spacy.load('en_core_web_sm')

        # ensure a sentencizer exists (works without the full parser)
        if 'sentencizer' not in nlp.pipe_names:
            try:
                nlp.add_pipe('sentencizer')
            except Exception:
                # ignore if adding fails for any reason
                pass

        doc = nlp(text)
        sents = [sent.text.strip() for sent in doc.sents if sent.text.strip()]
        if sents:
            return sents

        # fallback simple regex split if spaCy produced no sentences
        import re
        parts = [s.strip() for s in re.split(r'(?<=[.!?])\s+', text) if s.strip()]
        return parts if parts else [text]

    except Exception:
        # fallback to NLTK sentence tokenizer
        import nltk
        nltk.download('punkt', quiet=True)
        return nltk.sent_tokenize(text)

# example
sentence_tokenize("This is one sentence. And here is another! Is this a third? Yes.")
```

<span style="font-size:32px">Stemming</span>

```python
def stemmer(text):
    if not isinstance(text , str):
        return text
    from nltk.stem import PorterStemmer
    ps = PorterStemmer()

    return ' '.join([ps.stem(word) for word in text.split()])

stemmer("walked walking walk ")
```
