tf-idf = Term Frequency - Inverse Document Frequecy

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
data = {
    "text" : [
        "Naruto better onepiece",
        "onepiece better Bleach",
        "Bleach better AOT",
        "AOT better Naruto",
        "Blueloch better demonslayer",
        "demonslayer bluelock rezero demonslayer"
    ],
    "label" : [1 , 0 , 1, 1, 0 , 0]
}

df = pd.DataFrame(data)
df
```

```python
tf = TfidfVectorizer()
idf = tf.fit_transform(df['text'])
```

```python
idf.toarray()
```

```python
tf.idf_
```
