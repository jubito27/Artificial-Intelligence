```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
```

```python
data = {
    "text" : [
        "Naruto better onepiece",
        "onepiece better Bleach",
        "Bleach better AOT",
        "AOT better Naruto",
        "Blueloch better demonslayer",
        "demonslayer bluelock rezero demonslayer"
    ],
    "label" : [1 , 0 , 1, 1, 0 , 0]
}

df = pd.DataFrame(data)
df
```

```python
cv = CountVectorizer(ngram_range = (3,3))
bow = cv.fit_transform(df['text'])
bow.toarray()
```

```python
print(len(cv.vocabulary_))
cv.vocabulary_
```
